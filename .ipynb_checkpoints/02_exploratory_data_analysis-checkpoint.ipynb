### Correlation Analysis

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    credit_data.corr(numeric_only=True),
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
corr_target = (
    credit_data
    .corr(numeric_only=True)["target"]
    .sort_values(ascending=False)
)

plt.figure(figsize=(6,10))

sns.heatmap(
    corr_target.to_frame(),
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Correlation of Features with Target")
plt.show()
```

###### Correlation	Meaning	Interpretation
###### +1   ----->    Perfect positive	    ------>     As one variable increases, the other always increases.
###### +0.8 ----->    Strong positive	    ------>     Strong tendency to increase together.
###### 0    ------>   No linear relationship  ------>  One variable doesn't tell you much about the other.
###### -0.8   ----->  Strong negative	  ------>       One increases while the other decreases.
###### -1	----->    Perfect negative	    ------>     They move exactly opposite.  

###### - Repayment history variables (pay_1–pay_6) show the strongest positive correlation with default, suggesting that customers with recent payment delays are more likely to default.
###### - Monthly bill amounts are highly correlated with one another, indicating that customer outstanding balances tend to remain consistent across consecutive months.
###### - Monthly payment amounts also exhibit positive correlations, reflecting consistent payment behavior over time.
###### - Credit limit has a weak negative correlation with default, suggesting that customers with higher credit limits tend to have a slightly lower default risk.

### Credit Utilization

##### It measures how much of the available credit limit a customer is using.

```python
credit_data["credit_utilization"] = (
    credit_data["average_bill"] / credit_data["limit_bal"]
)
```

```python
sns.boxplot(
    data=credit_data,
    x="target",
    y="credit_utilization"
)
```

##### Boxplot shows:

##### Target = 0 (No Default) → Lower median utilization
##### Target = 1 (Default) → Higher median utilization
##### Customers who use a larger proportion of their available credit are more likely to default.

```python
utilization = (
    credit_data
    .groupby("pay_1")["credit_utilization"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(8,5))

sns.lineplot(
    data=utilization,
    x="pay_1",
    y="credit_utilization",
    marker="o"
)

plt.title("Average Credit Utilization by Repayment Status")
plt.xlabel("Repayment Status")
plt.ylabel("Average Credit Utilization")

plt.show()
```

```python
credit_data["credit_utilization"].describe(percentiles=[0.90, 0.95, 0.99])
```

###### The median credit utilization is 28.5%, while the average utilization is 37.3%, indicating a positively skewed distribution. 
###### Approximately 99% of customers maintain utilization below 105% of their credit limit, with only a small number of customers exceeding their available credit. 
###### These extreme cases likely represent financially distressed customers and warrant further investigation.

### Payment Ratio

##### Measures how much of the bill the customer actually pays.

```python
credit_data["average_payment"] = credit_data[
    [
        "pay_amt1",
        "pay_amt2",
        "pay_amt3",
        "pay_amt4",
        "pay_amt5",
        "pay_amt6"
    ]
].mean(axis=1)
```

```python
credit_data["payment_ratio"] = np.where(
    credit_data["average_bill"] > 0,
    credit_data["average_payment"] / credit_data["average_bill"],
    np.nan
)
```

```python
credit_data[["average_bill", "average_payment"]].describe()
```

```python
credit_data["payment_ratio"].describe(
    percentiles=[0.25,0.5,0.75,0.9,0.95,0.99]
)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    credit_data["payment_ratio"],
    bins=50
)

plt.title("Distribution of Payment Ratio")
plt.xlabel("Payment Ratio")
plt.ylabel("Customers")

plt.show()
```

```python
#inspecting outliers 

credit_data.loc[
    credit_data["payment_ratio"] > 10,
    ["average_bill", "average_payment", "payment_ratio", "target"]
].sort_values("payment_ratio", ascending=False)
```

##### Here I've engineered the payment ratio using average payment divided by average bill. It has been observed that the extreme ratios are
##### caused by customers with very small average bills but normal payment amounts. To better align payments with their corresponding billing cycles, I
##### revised the feature to compute month-specific payment ratios (pay_amt_i / bill_amt_i) and then aggregated them.

```python
ratio_cols = []

for i in range(1, 7):
    col = f"payment_ratio_{i}"
    credit_data[col] = np.where(
        credit_data[f"bill_amt{i}"] > 0,
        credit_data[f"pay_amt{i}"] / credit_data[f"bill_amt{i}"],
        np.nan
    )
    ratio_cols.append(col)

credit_data["avg_payment_ratio"] = credit_data[ratio_cols].mean(axis=1)
```
