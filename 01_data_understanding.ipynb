```python
# Make the first row the column names
credit_data.columns = credit_data.iloc[0]

# Remove the first row since it's now the header
credit_data = credit_data.iloc[1:].reset_index(drop=True)

# Check the result
credit_data.head()
```

# 

```python
credit_data.groupby("sex")["target"].value_counts(normalize=True).mul(100).round(2)
#male = 1, female = 2 ==> males defaulting slightly more than females
```

```python
credit_data.groupby("education")["target"].value_counts(normalize=True).mul(100).round(2)
# (1 = graduate school; 2 = university; 3 = high school; 4 = others)
#high school graduates have the highest defeault rate
```

```python
credit_data.groupby("marriage")["target"].value_counts(normalize=True).mul(100).round(2)
# (1 = married; 2 = single; 3 = others)
# singles default rate is less
```

```python
# creating age groups
bins = [20, 30, 40, 50, 60, credit_data["age"].max() + 1]
labels = ["20s", "30s", "40s", "50s", "60+"]

credit_data["age_group"] = pd.cut(
    credit_data["age"],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
credit_data["age_group"].value_counts()
```

```python
credit_data.groupby("age_group")["target"].value_counts(normalize=True).mul(100).round(2)
#highest default rate is seen in 60+ age group bit it has significantly lower counts
# lowest default rate seen in 30s age group
```
